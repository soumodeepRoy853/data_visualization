## Basic Matplotlib operations

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11) # X axis
x
```

```python
y = x ** 2 # Y axis
y
```

```python
plt.plot(x, y)
plt.title("Line Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

## Subplots in Matplotlib

```python
# Create Subplots
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(y,x)

plt.subplot(2,2,3)
plt.plot(x**2, y)

plt.subplot(2,2,4)
plt.plot(y**2, x)
```

## Object Oriented Way

```python
# It is used for controll on plot figure
# It is used for create multiple plots on same figure
fig = plt.figure() # Axis1 plot
axis1 = fig.add_axes([0.1, 0.1, 0.9, 0.9]) # left, top, right, bottom
axis1.plot(x,y)
plt.title("Figure 1")
plt.xlabel("X axis")
plt.ylabel("Y axis")


#Now create another plot inside axis1 plot
# Axis2 plot inside Axis1 plot
axis2 = fig.add_axes([0.2, 0.3, 0.2, 0.2]) # left, top, right, bottom
axis2.plot(y,x)
plt.title("Figure 2")
plt.xlabel("X axis")
plt.ylabel("Y axis")

axis3 = fig.add_axes([.2, .7, 0.2, 0.2]) # left, top, right, bottom
axis3.plot(x, y)
plt.title("Figure 3")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

## Types of Plots

```python
plt.scatter(x,y) # Scatter Plot
```

```python
# Histogram plot
plt.hist(y)
```

## Save Plot

```python
fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.5, 0.5])
ax.plot(x, y)
plt.savefig("Basicplot.png")
```
