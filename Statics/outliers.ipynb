```python
arr = np.array([1, 2, 3, 5, 7, 9, 12, 14, 16, 18, 22, 32, 26, 23, 101])
arr
```

```python
# Create Quartiles1 (Q1)
Q1 = np.percentile(arr, 25)
Q1
```

```python
# Create Quartiles2 (Q2)
Q2 = np.percentile(arr, 50)
Q2
```

```python
# Create Quartiles3 (Q3)
Q3 = np.percentile(arr, 75)
Q3
```

```python
# Create IQR
IQR = Q3 - Q1
IQR
```

```python
# Create Upper Fench
UF = Q3 + (1.5 * IQR)
UF
```

```python
# Create Lower Fench
LF = Q1 - (1.5 * IQR)
LF
```

```python
# Remove all the numbers above UF
l = []
for i in arr:
    if i>= LF and i <= UF:
        l.append(i)

# Add into new array
arr2 = np.array(l)
arr2
```

```python
# Create Box plot using val
import seaborn as sns
```

```python
# Create a boxplot with outlier
sns.boxplot(x = arr)
```

```python
# Create a boxplot without outlier
sns.boxplot(x = arr2)
```
