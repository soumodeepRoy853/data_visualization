## Categorical Plots
Now Let's discuss using seaborn to plot categorical data.
There are a few main plot types for this:
1. boxplot
2. violinplot
3. stripplot
4. swarmplot
5. barplot
6. countplot

```python
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
df
```

## Countplot

```python
sns.countplot(x = df["sex"])
```

```python
sns.countplot(x = df["sex"], hue=df["smoker"])
```

## Barplot

```python
# By default estimator is mean of total_bill for sex
sns.barplot(x=df["sex"], y=df["total_bill"])
```

```python
# Count total sum of tip for by sex using estimator
import numpy as np
sns.barplot(x=df["sex"], y=df["tip"], estimator=np.sum)
```

## Boxplot

```python
sns.boxplot(x="tip", y="day", data=df, hue="sex", palette="rainbow")
```

## Violin Plot

```python
sns.violinplot(x=df["day"], y=df["total_bill"], data=df)
```

## Strip Plot

```python
sns.stripplot(x="day", y="total_bill", data=df)
```

## Swarm Plot

```python
sns.swarmplot(x="day", y="total_bill", data=df)
```

```python
# Swarm Plot and Violin Plot in same canvas
sns.violinplot(y="day", x="total_bill", data=df)
sns.swarmplot(y="day", x="total_bill", data=df)
```
