## Distribution Plot
It is use to visualize the distribution of quantitative data.

Let's discuss some distribution plots:
1. distplot/histplot
2. jointplot
3. pairplot
4. rugplot


```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load seaborn dataset
df = sns.load_dataset("tips")
df
```

```python
df["size"].unique()
```

## 1. Distplot/HistPlot

```python
# Create Hist Plot/Dist Plot
sns.histplot(df['total_bill'], bins=20)
```

```python
# Draw Karnel Density Estimation(KDE)
sns.histplot(df['total_bill'], bins=20, kde=True)
```

```python
# Draw two plot using subplot
plt.subplot(1, 2, 1)
sns.histplot(df['total_bill'], bins=20, kde=True)

plt.subplot(1,2,2)
sns.histplot(df['tip'], bins=20, kde=True)
```

## 2. JointPlot

```python
# Create Joint Plot (Total Bill + Tip)
sns.jointplot(x="total_bill", y="tip", data=df)
```

```python
# Initially it is a scatter plot but we can change it into others plot by using kind="plot name"
sns.jointplot(x="total_bill", y="tip", data=df, kind="reg") # kind:'scatter', 'kde', 'hist', 'hex', 'reg', 'resid'
```

## 3. Pair Plot

```python
# Create Pair plot
sns.pairplot(df)
```

```python
# Use Hue to map plot aspects to different colors
sns.pairplot(df, hue="sex")
```

```python
# Palette: Set of colors for mapping the hue variable.
sns.pairplot(df, hue="sex", palette="rainbow")
```

## 4. Rug Plot

```python
sns.rugplot(x="total_bill", y="tip", data=df, hue="sex")
```
